```python
import mlflow
from mlflow.models import infer_signature

import pandas as pd

from sklearn import datasets
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
```

```python
### load the dataset
X , Y = datasets.load_iris(return_X_y=True)

### split the dataset into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=7)

### Define model parameters
params = {'penalty':'l2' , 'solver':'lbfgs', 'max_iter':1000, 'random_state':7}

### Train the model
model = LogisticRegression(**params)
model.fit(X_train, Y_train);
```

```python
#inference model
Y_pred = model.predict(X_test)
Y_pred
```

```python
accuracy = accuracy_score(Y_test,Y_pred)
precision = precision_score(Y_test,Y_pred,average='weighted')
recall = recall_score(Y_test,Y_pred,average='weighted')
f1 = f1_score(Y_test,Y_pred,average='weighted')

metrics = {'accuracy':accuracy,'precision':precision,'recall':recall,'f1':f1}
```

```python
X_train[:10],model.predict(X_train[:10])
```

```python
### Track results
mlflow.set_tracking_uri(uri="http://127.0.0.1:5000")

### create a new experiment
mlflow.set_experiment("MLOPS-Project-1")

### start an mlflow start
with mlflow.start_run():
    # log parameters params should be key value pairs if you are passing multiple metrics otherwise use param instead of params
    mlflow.log_params(params)

    # log metrics
    mlflow.log_metrics(metrics)

    # set tag
    mlflow.set_tag('Training info', 'Basic LR model on Iris dataset')

    # infer the model signature
    schema = infer_signature(X_train[:10],model.predict(X_train[:10]))           # stores the schema of the data given to the model input and output. it takes a subset of the training data as input and the prediction of this subset in the output

    # log the model
    model_info = mlflow.sklearn.log_model(
        sk_model=model,
        artifact_path="Iris-model",
        signature=schema,
        input_example=X_train[:10],
        registered_model_name='tracking-gettingstarted'

    )
```
